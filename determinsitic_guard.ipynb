## NO LLM, plain if-else

```python
harmful_keywords = ['hack' , 'delete' , 'bomb']
def deterministic_guard(text : str) -> bool :
    return any(kw in text.lower() for kw in harmful_keywords)
```

```python
test_inputs = [
    "how to hack system?" , "what is myt age?" , "how to create bomb?"
]

for inp in test_inputs :
    blocked = deterministic_guard(inp)
    status = "blocked" if blocked else "allowed"
    print(f"status : {status} \n")
```
